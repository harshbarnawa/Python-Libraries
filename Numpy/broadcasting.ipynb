## Broadcasting in numpy array

```python
import numpy as np
```

```python
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000], 
    [1, 120000, 140000, 160000, 190000], 
    [1, 200000, 230000, 260000, 300000], 
    [1, 180000, 210000, 240000, 270000], 
    [1, 160000, 185000, 205000, 230000] 
])
```

```python
monthly_avg = sales_data[:,1:] /12
print(monthly_avg)

## this scalar number streched for the vector operation
## it's called broadcasting it implements at every position of vector element
## individually
```
