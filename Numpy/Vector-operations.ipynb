## Vector operations

```python
import numpy as np
```

```python
vec1 = np.array([1, 2, 3, 4, 5])
vec2 = np.array([6, 7, 8, 9, 10])

print(("vector addition: ", vec1 + vec2))
print(("vector multiplication: ", vec1 * vec2))
print("Dot product", np.dot(vec1, vec2)) ## Dot product is the sum of the products of corresponding elements of two vectors


## angle b/w vecs
angle = np.arccos(np.dot(vec1, vec2)/ (np.linalg.norm(vec1) * np.linalg.norm(vec2)))
print(angle)
```

```python
## vectorized operations
restaurant_types = np.array(['biryani', 'chinese', 'pizza', 'burger', 'cafe'])

vectorized_upper = np.vectorize(str.upper) ## creating vectorized method
print("Vectorized upper", vectorized_upper(restaurant_types))
```
