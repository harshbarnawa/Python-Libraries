## Advance operation with business examples

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000], # Paradise Biryani
    [2, 120000, 140000, 160000, 190000], # Beijing Bites
    [3, 200000, 230000, 260000, 300000], # Pizza Hub
    [4, 180000, 210000, 240000, 270000], # Burger Point
    [5, 160000, 185000, 205000, 230000] # Chai Point
])

print("==== Zomato sales analysis ====")
print("\n Sales data shape", sales_data.shape)
print("\n First three \n", sales_data[0:3]) # basic idea
print("\n Selection \n", sales_data[:, 1:]) # first colon selects whole 2nd is range

print("\n")
print(np.sum(sales_data, axis=0))
year_total = np.sum(sales_data[:, 1:], axis=0)
print(year_total)

# Minimum sales per restaurant
min_sales = np.min(sales_data[:, 1:], axis=1)

print("\nMin sales per rest:",min_sales)

# max sales/year

max_sales = np.max(sales_data[:, 1:], axis = 0)
print("Max sales per year:",max_sales)

# Average sales per rest
avg_sales = np.mean(sales_data[:,1:], axis = 1)
print("avg sales per resta", avg_sales)
```

```python
cumsum = np.cumsum(sales_data[:,1:], axis=1)
print(cumsum) ## cumulative sales # adding per year

plt.figure(figsize=(10, 6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average cumulative sales across all restaurant ")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
